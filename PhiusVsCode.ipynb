```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd. read_csv("C:/Users/amitc_crl/Downloads/Modeling Matrix - Results.csv")

df.head()

# create data
x = list(df['Code Version'])
MELs = np.array(list(df['MELs[kBtu/yr]']))
lights = np.array(list(df['Lighting [kBtu/yr]']))
Appliances = np.array(list(df['Appliances [kBtu/yr]']))
AuxEnerg = np.array(list(df['Aux Energy [kBtu/yr]']))
SHW = np.array(list(df['SHW [kBtu/yr]']))
cooling= np.array(list(df['Space Cooling [kBtu/yr]']))
heating =  np.array(list(df['Space Heating [kBtu/yr]']))

# plot bars in stack manner
plt.bar(x, MELs, color='xkcd:navy')
plt.bar(x, lights, bottom=MELs, color='xkcd:yellow')
plt.bar(x, Appliances, bottom=MELs+lights, color='xkcd:nasty green')
plt.bar(x, AuxEnerg, bottom=MELs+lights+Appliances, color='xkcd:medium grey')
plt.bar(x, SHW, bottom=MELs+lights+Appliances+AuxEnerg, color='xkcd:purple')
plt.bar(x, cooling, bottom=MELs+lights+Appliances+AuxEnerg+SHW, color='xkcd:teal blue')
plt.bar(x, heating, bottom=MELs+lights+Appliances+AuxEnerg+SHW+cooling, color='xkcd:brick red')
plt.xlabel("Standard")
plt.ylabel("Site Energy [kBtu/yr]")
plt.legend(["MELs", "Lights", "Appliances", "Aux Energy", "DHW", 'Cooling', 'Heating'], loc='upper right', fontsize=8)
plt.suptitle("Phius 2021 CORE vs IECC 2021")
plt.title("1A | Single Family | Miami Florida")
plt.ylim((0, 1.1*((MELs[0]+lights[0]+Appliances[0]+AuxEnerg[0]+SHW[0]+cooling[0]+heating[0]))))
plt.show()
```
