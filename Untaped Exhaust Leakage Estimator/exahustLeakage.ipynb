```python
import numpy as np
```

```python
# Determine barometric pressure in hPa at a given altitude using the barometric formula
H = 362


h = H * 0.3048  # Convert altitude from feet to meters

P_0 = 101325  # Sea level standard atmospheric pressure in Pascals
T_0 = 288.15  # Standard temperature at sea level in Kelvin
g = 9.80665  # Gravitational acceleration in m/s^2
L = 0.0065  # Temperature lapse rate in K/m
R = 8.31447  # Universal gas constant in J/(mol·K)
M = 0.0289644  # Molar mass of Earth's air in kg/mol

P_barometric = P_0 * (1 - (L * h) / T_0) ** (g * M / (R * L))

P_barometric_IP = P_barometric * 0.40146307866177

rho_SI = P_barometric / (287.05 * (T_0 - L * h))  # Density of air in kg/m^3
```

```python
C_d = 0.61  # Discharge coefficient
A = 0.9  # Cross-sectional area in ft^2
dP = 0.2000732
rho_IP = rho_SI * 0.06243  # Density of air in lb/ft^3

Q = 776 * C_d * A * ((2 * dP) / rho_IP) ** 0.5

print(str(Q) + " SCFM")
```

```python
A_env = 6772 # Envelope Area ft^2
BD_target = 0.06  # Target building leakage in CFM per ft^2 at 0.2 inH2O
D = 8 #  Duct Diameter in inches
L = 100  #  Equivalent Length in ft
d = (D/12)  # Diameter in ft
r = d / 2  # Radius in ft

e = 0.0005  # Roughness in ft

A = 3.142 * r**2  # Cross-sectional area in ft^2

dP_Duct = 0

# V = 748  # Velocity in ft/min

# Q = V * A  # Flow rate in SCFM

Q = 1  # Initial flow rate in SCFM

while dP_Duct < 0.2:
    V = Q / A  # Velocity in ft/min

    Re = 8.5 * D * V # Reynolds number

    Haaland = (-1.8 * np.log10( ( ( (3.243 * e/D) )**1.11 ) + (6.9/Re) ))
    f = (1 / Haaland ** 2)

    dP_Duct = (12 * f * (L/D) * (rho_IP) * (V/1097) ** 2)

    Q = Q + 1

envelope_leakage = A_env * BD_target  # Leakage through envelope in SCFM

print("Additional leakage through duct = " + str (Q) + " SCFM at " + str(dP_Duct) + " inH2O pressure drop")
print("Taped Test envelope leakage = " + str(envelope_leakage) + " SCFM")
print("Total Leakage = " + str(Q + envelope_leakage) + " SCFM @ 0.2 inH2O pressure drop")
print("WUFI / METrun input leakage = " + str((Q + envelope_leakage)/A_env) + " CFM per ft^2 at 0.2 inH2O")
```
