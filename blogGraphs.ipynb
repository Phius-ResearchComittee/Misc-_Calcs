```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
filepath = "D:/Chicago/2/results/2_ResultsTable.csv"

df = pd.read_csv(filepath)

df.head()
```

```python
ax = df.plot(kind='bar', x='Year', y=['Phius Prescriptive_EUI','IECC 2021_EUI'], rot=90, figsize=(12,4), title='EUI [kBtu/sf yr]')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
ax = df.plot(kind='bar', x='Year', y=['Phius Precriptive_Danger Hours','IECC 2021_Danger Hours'], rot=90, figsize=(12,4), title='Heat Index Hours in Danger')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
